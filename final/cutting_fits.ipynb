```python
from astropy.io import fits
import numpy as np
from matplotlib import pyplot as plt

import glob
from tqdm import tqdm
```

```python
|def remove_fits_layers(image_hdu, layers):
    n_layers = len(image_hdu.data)
    new_data = image_hdu.data[layers]
    new_header = image_hdu.header.copy()

    new_cards = []
    for idx, layer in enumerate(layers):
        new_cards.append((f'TYPE{idx}', new_header[f'TYPE{layer}'], new_header.comments[f'TYPE{layer}']))

    for i in range(n_layers):
        new_header.remove(f'TYPE{i}', ignore_missing=True)

    for key, val, comment in new_cards:
        new_header[key] = (val, comment)

    new_header['NAXIS3'] = len(layers)

    return fits.PrimaryHDU(new_data, new_header)
```

```python
def cut_fits(input_path, output_path):
    image_hdu = fits.open(input_path)[0]
    return remove_fits_layers(image_hdu, [16, 2, 7]).writeto(output_path)
```

```python
input_path = ''
output_path = ''
names = glob.glob(input_path + '\\*.fits')
```

```python
for name in tqdm(names[:30]):
    fname = name.split('\\')[-1]
    cut_fits(name, output_path + '\\' + fname)
```
