```python
import numpy as np
from matplotlib import pyplot as plt

from mpl_toolkits.mplot3d import Axes3D
```

```python
def Lorenz_system(x_vector):
    x, y, z = x_vector.T
    sigma = 10
    rho = 28
    beta = 8./3.

    dx = sigma*(y-x)
    dy = x*(rho-z)-y
    dz = x*y-beta*z

    return np.array([dx, dy, dz]).T
```

```python
n = 100000
dt = 40./n
```

```python
x_vector = np.array([5,5,5], dtype='float').T


trajectory = [x_vector.T.tolist()]

for t in range(n):
    x_vector += dt*Lorenz_system(x_vector)
    trajectory.append(x_vector.T.tolist())
```

```python
points = np.array(trajectory).T
fig = plt.figure()
ax = fig.add_subplot(111, projection = '3d')
ax.plot(points[0], points[1], points[2])
ax.scatter(*points.T[0], color = 'red')
```
